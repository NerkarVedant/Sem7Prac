# LP3 Machine Learning Practical 4 - Gradient Descent

**Problem statement:** Implement Gradient Descent Algorithm to find the local minima of a function. For example, find the local minima of the function y = (x+3)^2 starting from the point x = 2.

```python
import matplotlib.pyplot as plt

# Function and its derivative
def f(x):
    return (x + 3)**2

def df(x):
    return 2 * (x + 3)

# Initialize parameters
x_old = 2          # Starting point
alpha = 0.1        # Learning rate
precision = 0.00001
max_iters = 100

# Lists for visualization
x_list = [x_old]
y_list = [f(x_old)]

# Gradient Descent Loop
for i in range(max_iters):
    grad = df(x_old)
    x_new = x_old - alpha * grad

    x_list.append(x_new)
    y_list.append(f(x_new))

    if abs(x_new - x_old) < precision:
        break

    x_old = x_new

# Display results
print(f"Local minimum occurs at x = {x_new:.5f}")
print(f"Minimum value of function y = {f(x_new):.5f}")

# Visualization
plt.figure(figsize=(10,5))
x_vals = [i for i in range(-10, 5)]
y_vals = [f(i) for i in x_vals]
plt.plot(x_vals, y_vals, label='y = (x+3)^2', color='blue')
plt.scatter(x_list, y_list, color='red', label='Descent Steps')
plt.title('Gradient Descent Convergence')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

```python
# =====================================================================
# THEORY AND EXPLANATION - Practical 4: Gradient Descent
# =====================================================================
#
# 1. PROBLEM
# Find the local minimum of y = (x + 3)^2 starting from x = 2, using the Gradient
# Descent algorithm.
#
# 2. KEY THEORY
# a) Gradient Descent
#    An iterative optimisation algorithm that finds a minimum of a function by
#    repeatedly stepping in the direction OPPOSITE to the gradient (slope), because
#    the gradient points towards the steepest increase.
#        x_new = x_old - alpha * f'(x_old)
#    alpha is the learning rate (step size). The same idea trains neural networks
#    and linear/logistic regression, where the function is the loss.
#
# b) The function used
#    f(x)  = (x + 3)^2
#    f'(x) = 2 * (x + 3)
#    Setting f'(x) = 0 gives x = -3, where f = 0. This is a convex function (a
#    parabola) so its only local minimum is also the global minimum.
#
# c) Learning rate
#    Too small: converges but needs many iterations.
#    Too large: overshoots the minimum, may oscillate or diverge.
#    For this function each step multiplies the distance (x + 3) by (1 - 2*alpha).
#    With alpha = 0.1 the factor is 0.8, so it converges smoothly. For alpha = 0.5
#    it reaches -3 in a single step. For alpha > 1 the factor has magnitude above 1
#    and the method diverges.
#
# d) Stopping criteria
#    Stop when the change in x is smaller than the chosen precision (0.00001), or
#    when max_iters (100) is reached, whichever comes first.
#
# 3. STEPS (code flow)
#    1. Define f(x) and its derivative df(x).
#    2. Set the start point x = 2, alpha = 0.1, precision = 0.00001, max 100 steps.
#    3. Loop: compute the gradient, update x_new = x_old - alpha * gradient, store
#       the point for plotting.
#    4. If |x_new - x_old| < precision, stop. Otherwise set x_old = x_new.
#    5. Print the final x and f(x), plot the curve and the red descent points.
#
# 4. DRY RUN (first steps)
#    x0 = 2     gradient = 2*(2+3)     = 10    x1 = 2 - 0.1*10     = 1.0
#    x1 = 1.0   gradient = 2*(1+3)     = 8     x2 = 1 - 0.1*8      = 0.2
#    x2 = 0.2   gradient = 2*(0.2+3)   = 6.4   x3 = 0.2 - 0.1*6.4  = -0.44
#    The distance from -3 shrinks by a factor of 0.8 every step (5, 4, 3.2, 2.56, ...)
#
# 5. RESULT AND HOW TO READ IT
#    Local minimum at x = -2.99996 and y = 0.00000 (the exact answer is x = -3).
#    The loop stops after 53 iterations, when the step size falls below 0.00001.
#    In the plot the red points start at x = 2 (y = 25) and move down the left side
#    of the parabola. The steps get smaller and smaller near the bottom because the
#    gradient itself becomes smaller there.
#
# 6. VIVA QUESTIONS
#    Q: Why do we subtract the gradient?
#    A: The gradient points uphill; subtracting it moves us downhill.
#    Q: What happens if the learning rate is too high?
#    A: It overshoots and may diverge. For this function alpha > 1 diverges.
#    Q: Local versus global minimum?
#    A: A convex function has only one minimum, which is global. A non-convex
#       function can have several local minima and gradient descent can get stuck
#       in one of them.
#    Q: Where is gradient descent used in ML?
#    A: Training linear regression, logistic regression and neural networks by
#       minimising a loss function.
#    Q: Batch, stochastic and mini-batch gradient descent?
#    A: Batch uses all the data per step, stochastic uses one sample, mini-batch
#       uses a small group (the usual choice).
```
